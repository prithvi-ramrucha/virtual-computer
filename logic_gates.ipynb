# **Logic Gates**

Transistors are the most fundamental building blocks of modern computers. We can imagine them as tiny switches. Multiple transistors can be arranged together to form logic gates such as `AND` and `OR`. Once `NOR` or a `NAND` gates are created from these components, the resulting logic system is Turing complete, meaning that it can be used to construct any other Boolean logic gate and, given sufficient memory and time, can perform any computation that is computationally possible. In other words, a general purpose computer can be built. Our most basic units will be logic gates. Let us start with the most simple logic gate; the `NOT` gate:

```python
class NOT:
    def __init__(self, input):
        self.input = input

    def output(self):
        return not self.input
```

It simply inverts the input,

```python
gate = NOT(0)
print(gate.output())  # Output: False
```

Using transistors, the `NOT` gate can be constructed as follows,

We can create the other logic gates,

```python
class AND:
    def __init__(self, input1, input2):
        self.input1 = input1
        self.input2 = input2

    def output(self):
        return self.input1 and self.input2

class OR:
    def __init__(self, input1, input2):
        self.input1 = input1
        self.input2 = input2

    def output(self):
        return self.input1 or self.input2
```

In reality, the `NOR` and `NAND` gates are constructed using transistors. However, for our abstraction, we construct them using our existing logic gates. Since the `NOT` gate can be thought of as an inverter,

```python
class NAND:
    def __init__(self, input1, input2):
        self.input1 = input1
        self.input2 = input2

    def output(self):
        gate1 = AND(self.input1, self.input2)
        gate2 = NOT(gate1.output())
        return gate2.output()

class NOR:
    def __init__(self, input1, input2):
        self.input1 = input1
        self.input2 = input2

    def output(self):
        gate1 = OR(self.input1, self.input2)
        gate2 = NOT(gate1.output())
        return gate2.output()
```
